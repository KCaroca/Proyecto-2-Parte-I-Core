**Detección y Corrección de Incongruencias Categóricas**

Antes de aplicar cualquier técnica de limpieza, es importante revisar los datos para detectar posibles problemas. En este paso, se identificarán las columnas de texto y se analizarán sus valores únicos junto con su frecuencia. El objetivo es encontrar posibles inconsistencias, como espacios adicionales, errores de escritura o diferencias en el uso de mayúsculas y minúsculas, antes de realizar cualquier modificación en el dataset.

```python
cat_cols = df.select_dtypes(include='object').columns

for col in cat_cols:
    print(f"\n{'='*50}")
    print(f"Variable: {col}")
    print(f"Valores únicos: {df[col].nunique()}")
    print(df[col].value_counts().head(20))
```

Se realizó una revisión de las variables categóricas mediante el análisis de sus valores únicos y frecuencias de aparición, con el objetivo de identificar posibles inconsistencias de codificación. Los resultados muestran una estructura coherente en todas las variables analizadas, registrándose 25 departamentos, 196 provincias y 1.538 distritos, lo que refleja una amplia cobertura geográfica del territorio peruano. Asimismo, la variable Servicio Prestacional presenta 36 categorías correspondientes a distintos tipos de atención en salud, mientras que la variable CAPITA contiene únicamente dos categorías (S y N). En esta revisión no se observaron inconsistencias evidentes que comprometieran la calidad de los datos, aunque la alta cardinalidad de algunas variables geográficas deberá considerarse en etapas posteriores de modelamiento.

```python
categoricas = [
    'DEPARTAMENTO',
    'PROVINCIA',
    'DISTRITO',
    'SERVICIO_PRESTACIONAL',
    'CAPITA'
]

for col in categoricas:
    print(f"\n{col}")
    print(df[col].sort_values().unique()[:20])
```

Se realizó una inspección de los valores únicos presentes en las variables categóricas del conjunto de datos mediante su ordenamiento alfabético. Esta revisión permitió verificar la estructura y consistencia de las categorías asociadas a la ubicación geográfica (departamento, provincia y distrito), los servicios prestacionales y la condición de afiliación (CAPITA).

Los resultados muestran una nomenclatura homogénea y consistente entre las categorías observadas, sin evidenciar duplicidades aparentes derivadas de diferencias de escritura. Asimismo, se constató una amplia diversidad geográfica representada en los registros y una variedad de servicios de atención vinculados al control y seguimiento de la gestante. En el caso de la variable CAPITA, únicamente se registran las categorías "S" y "N", lo que confirma una clasificación binaria claramente definida.

Esta revisión permitió validar preliminarmente la calidad de las variables categóricas antes de aplicar procesos de estandarización y limpieza de datos.

```python
df_revision = df.copy()

for col in categoricas:
    df_revision[col] = (
        df_revision[col]
        .astype(str)
        .str.strip()
        .str.upper()
    )
```

```python
for col in categoricas:
    print(col)
    print("Antes:", df[col].nunique())
    print("Después:", df_revision[col].nunique())
    print("-"*30)
```

Con el propósito de verificar la existencia de inconsistencias asociadas al uso de mayúsculas, minúsculas o espacios en blanco, se aplicó un proceso de estandarización temporal sobre las variables categóricas. Posteriormente, se comparó la cantidad de categorías únicas antes y después de la transformación.

Los resultados muestran que el número de categorías permaneció sin cambios en todas las variables analizadas: Departamento (25), Provincia (196), Distrito (1.538), Servicio Prestacional (36) y Capita (2). Esto indica que no existían diferencias relevantes de formato que generaran categorías duplicadas o inconsistencias de codificación.

En consecuencia, se concluye que las variables categóricas presentan una estructura homogénea y una adecuada calidad de registro, permitiendo continuar con el análisis sin necesidad de realizar correcciones adicionales en esta etapa.

```python
numericas = ['EDAD','PESO','TALLA','EDAD_GESTACIONAL']

df[numericas].hist(
    figsize=(12,8),
    bins=30
)

plt.tight_layout()
plt.show()
```

Se analizaron las distribuciones de las variables numéricas mediante histogramas con el propósito de identificar patrones de comportamiento, posibles asimetrías y valores atípicos.

La variable EDAD presenta una distribución concentrada principalmente entre los 20 y 35 años, lo que es consistente con una población de mujeres en edad reproductiva. Por su parte, PESO muestra una distribución aproximadamente unimodal, con mayor frecuencia entre 50 y 80 kilogramos, aunque se observan algunos valores extremos que podrían corresponder a registros atípicos.

En el caso de TALLA, la mayor concentración de observaciones se encuentra entre 140 y 170 centímetros, rango compatible con valores antropométricos esperados para la población estudiada. Sin embargo, se observan algunos registros alejados del patrón general que deberán revisarse posteriormente mediante técnicas de detección de outliers.

La variable EDAD_GESTACIONAL presenta una distribución más dispersa, con predominio de registros entre las semanas 6 y 15 de gestación y una disminución gradual de la frecuencia hacia edades gestacionales mayores. Esta distribución refleja la presencia de controles prenatales realizados en distintas etapas del embarazo.

En términos generales, las distribuciones observadas son coherentes con la naturaleza del conjunto de datos. No obstante, la presencia de algunos valores extremos en variables antropométricas sugiere la necesidad de complementar el análisis mediante diagramas de caja (boxplots) para confirmar la existencia de posibles valores atípicos.

**Identificando Outliers**

```python

plt.figure(figsize=(12,8))

for i, col in enumerate(numericas, 1):
    plt.subplot(2,2,i)
    sns.boxplot(x=df[col])
    plt.title(col)

plt.tight_layout()
plt.show()
```

Los diagramas permitieron identificar la presencia de valores atípicos en las principales variables numéricas del conjunto de datos. En la variable EDAD, la mayoría de las observaciones se concentra entre los 20 y 35 años, aunque se detectan algunos registros extremos que alcanzan edades cercanas a los 100 años. En PESO se observa una distribución centrada entre 50 y 80 kg, junto con valores significativamente superiores a 100 kg e incluso algunos casos sobre 200 kg. Por su parte, la variable TALLA presenta registros mayoritariamente dentro de rangos esperables, aunque destaca un valor cercano a 1.150 cm, claramente inconsistente y probablemente asociado a un error de digitación.

La variable EDAD_GESTACIONAL muestra un comportamiento más estable, concentrándose principalmente entre las semanas 8 y 20, con pocos valores extremos cercanos a las 40 semanas, compatibles con embarazos a término.

En general, los valores atípicos más relevantes se encuentran en las variables PESO y TALLA, por lo que será necesario evaluarlos durante la etapa de preparación de datos para determinar si corresponden a observaciones válidas o errores de registro que puedan afectar los análisis posteriores.
