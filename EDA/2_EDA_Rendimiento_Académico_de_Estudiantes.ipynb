# **Proyecto 2 – Parte I (Rendimiento Académico de Estudiantes)**

**Rendimiento Académico de Estudiantes - Descripción del Dataset**

Este conjunto de datos recopila información académica, familiar, social y personal de estudiantes de enseñanza secundaria. Incluye antecedentes relacionados con características demográficas, nivel educativo de los padres, entorno familiar, hábitos de estudio, apoyo escolar, ausencias y desempeño académico, con el propósito de analizar los factores que influyen en el rendimiento escolar.

Fuente Kaggle: [Rendimiento Académico de Estudiantes](https://www.kaggle.com/datasets/beltranz88/estudiantes)

Cada fila representa a un estudiante e incorpora información sobre sus condiciones familiares, características personales, hábitos de estudio y resultados académicos obtenidos durante el año escolar.

Variables principales
- Edad
- Sexo
- Tipo de residencia (urbano/rural)
- Tamaño del grupo familiar
- Estado civil de los padres
- Nivel educativo de la madre
- Nivel educativo del padre
- Ocupación de la madre
- Ocupación del padre
- Tiempo de estudio semanal
- Apoyo familiar
- Apoyo escolar adicional
- Acceso a Internet en el hogar
- Estado de salud
- Número de ausencias
- Asignaturas reprobadas
- Nota primer parcial
- Nota segundo parcial
- Nota final (posible variable objetivo)

Estas variables permiten analizar cómo factores personales, familiares y académicos se relacionan con el desempeño final de los estudiantes.

Dimensiones del Dataset

El conjunto de datos contiene 1.044 registros y 33 variables, combinando variables numéricas y categóricas. La información disponible permite desarrollar análisis descriptivos, identificar factores asociados al rendimiento académico y construir modelos predictivos orientados a la clasificación o predicción del desempeño estudiantil.

```python
df_revision = df.copy()

for col in categoricas:
    df_revision[col] = (
        df_revision[col]
        .astype(str)
        .str.strip()
        .str.upper()
    )

for col in categoricas:
    print(col)
    print("Antes:", df[col].nunique())
    print("Después:", df_revision[col].nunique())
    print("-" * 30)
```

Se aplicó una normalización temporal de las variables categóricas para verificar si existían diferencias de formato que generaran categorías duplicadas. Al comparar la cantidad de valores únicos antes y después de la estandarización, se puede confirmar si las categorías originales eran consistentes.
```python
plt.figure(figsize=(8,5))
sns.histplot(df["nota_final"], bins=20, kde=True)

plt.title("Distribución de la Nota Final")
plt.xlabel("Nota Final")
plt.ylabel("Frecuencia")

plt.show()
```

La variable nota_final representa el resultado académico global del estudiante y constituye la principal candidata a variable objetivo. **Creación de variable rendimiento**

```python
df["rendimiento"] = pd.cut(
    df["nota_final"],
    bins=[-1, 9, 13, 20],
    labels=["BAJO", "MEDIO", "ALTO"]
)
```

```python
df["rendimiento"].isnull().sum()
```

```python
df["rendimiento"].value_counts()
```

Se creó la variable rendimiento a partir de la nota final, clasificando a los estudiantes en niveles bajo, medio y alto. Esta transformación permite convertir el problema en una tarea de clasificación, más adecuada para modelos basados en árboles de decisión. Se ajustaron los intervalos para incluir correctamente las notas iguales a cero dentro de la categoría de rendimiento bajo, evitando que esos registros quedaran sin clasificar.

**Distribución de la variable objetivo rendimiento**

```python
plt.figure(figsize=(8,5))

ax = sns.countplot(data=df, x="rendimiento")

plt.title("Distribución de la Variable Objetivo: Rendimiento Académico")
plt.xlabel("Nivel de Rendimiento")
plt.ylabel("Frecuencia")

for p in ax.patches:
    ax.annotate(
        f'{int(p.get_height())}',
        (p.get_x() + p.get_width() / 2, p.get_height()),
        ha='center',
        va='bottom'
    )

plt.show()
```

La distribución de la variable rendimiento permite observar cómo se reparten los estudiantes entre los niveles bajo, medio y alto. Este análisis es importante porque permite identificar si existe desbalance entre clases. Una distribución excesivamente concentrada en una sola categoría podría afectar el desempeño de modelos de clasificación, mientras que una distribución más equilibrada favorece el entrenamiento posterior.
